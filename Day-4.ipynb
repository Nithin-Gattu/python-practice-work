# operators

#### Arithematic operators

```python
a = 10
b = 3

print(a + b)    
print(a - b) 
print(a * b)  
print(a / b)   
print(a % b)    
print(a ** b)   
print(a // b)   
```

#### Comparision operators

```python
a = 10
b = 5

print(a == b)   
print(a != b)  
print(a > b)  
print(a < b)    
print(a >= b)
print(a <= b)   
```

#### Assignment Operators

```python
x = 20
print("Initial value:", x)
x = 45
print("=  :", x)
x += 5
print("+= :", x)
x -= 3
print("-= :", x)
x *= 2
print("*= :", x)
x /= 4
print("/= :", x)
x %= 3
print("%= :", x)
x //= 2
print("//=:", x)
x **= 3
print("**=:", x)
```

#### Logical Operators

```python
age = 20

print(age > 18 and age < 25)  
print(age < 18 or age > 30)    
print(not age > 18)            
```

#### Bitwise Operators

```python
a = 5 #101(binary code)
b = 3 #011(binary code)
print(a & b)   
print(a | b)   
print(a ^ b)    
print(~a)       
print(a << 1)   
print(a >> 1)  
```

#### Membership Operators

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)        
print("orange" in fruits)       
print("orange" not in fruits)   
print("banana" not in fruits)
```

#### Identity Operators

```python
a = [10, 20, 35]
b = a 
print(a is b) # b doesn't create a new list. Both a and b point to the same list (have same object)   
c = [10, 20, 35]        
print(a is c) # both lists contain the same values, but they are two different list objects.  
print(a is not c)
```
