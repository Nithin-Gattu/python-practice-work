## Datatypes

#### Numeric datatype

```python
age = 22
cgpa = 8.9
x = 3+5j
print(type(age))
print(type(cgpa))
print(type(x))
```

#### Sequence datatype

```python
#lists
lst_1 = [1,2,3,4,5]
print(lst_1)
lst_1[2] = 15 #mutable so can update the data
print(lst_1)
lst_2 = [10,20,30,"Student","name"]
print(lst_2)
print(type(lst_1))
print(type(lst_2))
```

```python
#tuple 
t = (1,2,3,4,5)
print(t)
print(type(t))
t(2) = 15 #immutable so cant update the original data
print(t)
```

```python
#strings
s = "college"
s1 = "a" #single letter(character) also termed as strings in python
print(s1)
print(s)
print(s[0])
print(type(s))
print(type(s1))
```

#### Mapping datatype

```python
#dictionary
d = {"ramesh":101,
     "suresh":102,
     "naresh":103}
print(d)
print(d["suresh"])
print(type(d))
```

```python
#sets
s1 = {1,2,3,4,5}
s2 = {1,2,2,3,4,3,5,1,}
print(s1)
print(s2)
print(type(s1))
print(type(s2))
```

#### Typecasting

```python
#Implicit casting
x = 100
y = 9.5
print(x)
print(y)
print(type(x),type(y))
z = x+y
print(z)
print(type(z))
```

```python
#Explicit casting
s = "python"
print(s)
print(type(s))
lst = list(s)
print(lst)
print(type(lst))
```
