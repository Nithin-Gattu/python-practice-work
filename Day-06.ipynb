### Lists

```python
numbers = [10, 20, 30, 40]
print(numbers)

#Multiple inputs at a time
numbers = list(map(int, input().split()))
print(numbers)     

#With different dataype
data = [10, "Python", 25.5, True]
print(data)
```

### List Operations

#### Concatenation

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
print(list1 + list2)
```

#### Repetition

```python
numbers = [1, 2, 3]
print(numbers * 2)
```

#### Membership

```python
numbers = [10, 20, 30, 40]
print(20 in numbers)
print(50 not in numbers)
print(45 in numbers)
```

### List methods

#### append()

```python
numbers = [10, 20, 30]
numbers.append(40)
print(numbers)
```

#### extend()

```python
numbers = [10, 20]
numbers.extend([30, 40, 50])
print(numbers)
```

#### insert()

```python
numbers = [10, 20, 40]
numbers.insert(2, 30)
print(numbers)
```

#### pop()

```python
numbers = [10, 20, 30, 40]
numbers.pop(1)
print(numbers)
```

#### remove()

```python
numbers = [10, 20, 30, 20]
numbers.remove(20) #removes the first matching values
print(numbers)
```

#### sort()

```python
#Ascending order
numbers = [40, 10, 30, 20]
numbers.sort()
print(numbers)

#Descending order
numbers.sort(reverse=True)
print(numbers)
```

#### count()

```python
numbers = [10, 20, 20, 30, 20]
print(numbers.count(20))
```

#### reverse()

```python
numbers = [10, 20, 30, 40]
numbers.reverse()
print(numbers)
```

#### index()

```python
numbers = [10, 20, 30, 40]
print(numbers.index(30))
```

#### copy()

```python
numbers = [10, 20, 30]
new_numbers = numbers.copy()
print(new_numbers)
```

#### clear()

```python
numbers = [10, 20, 30]
numbers.clear()
print(numbers)
```

### Built-in Functions

#### len()

```python
numbers = [10, 20, 30, 40, 50]
print(len(numbers))
```

#### min()

```python
numbers = [10, 5, 30, 20, 15]
print(min(numbers))
```

#### max()

```python
numbers = [10, 5, 30, 20, 15]
print(max(numbers))
```

#### sum()

```python
numbers = [10, 20, 30, 40]
print(sum(numbers))
```

#### sorted()

```python
numbers = [40, 10, 30, 20]
print(numbers)
print(sorted(numbers))
```
