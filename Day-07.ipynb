### Tuple

```python
numbers = (10, 20, 30, 40)
print(numbers)
print(type(numbers))
```

```python
#tuple with different datatype
data = (10, "Nithin", 25.5, True)
print(data)
```

```python
#tuple with one element
numbers = (10,)
print(numbers)
print(type(numbers))

#without comma
numbers = (10)
print(type(numbers))
```

### Tuple methods

#### Count()

```python
numbers = (10, 20, 20, 30, 20, 40)
print(numbers.count(20))
```

#### Index()

```python
numbers = (10, 20, 30, 40)
print(numbers.index(30))
```

#### Modify Tuple

```python
numbers = (10, 20, 30) #as immutable
numbers[1] = 50 
```

```python
#converting to list
numbers = (10, 20, 30)
numbers_list = list(numbers)
numbers_list[1] = 50
numbers = tuple(numbers_list)
print(numbers)
```

```python
#adding an element 
numbers = (10, 20, 30)
numbers_list = list(numbers)
numbers_list.append(40)
numbers = tuple(numbers_list)
print(numbers)
```

```python
#removing an element
numbers = (10, 20, 30, 40)
numbers_list = list(numbers)
numbers_list.remove(30)
numbers = tuple(numbers_list)
print(numbers)
```
