### STRINGS

```python
name = "Nithin"
city = 'Hyderabad'
message = """Welcome to Python"""

print(name)
print(city)
print(message)
```

#### Indexing

```python
#Positive Indexing
text = "Python"

print(text[0])
print(text[1])
print(text[2])
print(text[5])

#Negative Indexing
text = "Python"

print(text[-1])
print(text[-2])
print(text[-6])
```

#### Slicing

```python
text = "Python"

print(text[0:3])
print(text[2:5])
print(text[:4])
print(text[2:])

#Using step
text = "Python"

print(text[0:6:2])

#reverse a string
text = "Python"

print(text[::-1])
```

### String Operations

#### Concatenation

```python
first = "Gattu"
last = "Nithin"

name = first + " " + last

print(name)
```

#### Repetition

```python
text = "Python "
print(text * 3)
```

#### Comparison

```python
a = "apple"
b = "banana"

print(a == b)
print(a != b)
print(a < b)
print(a > b)
```

### String Methods

#### Case Sensitive

#### lower()

```python
text = "PYTHON"
print(text.lower())
```

#### upper()

```python
text = "python"
print(text.upper())
```

#### capitalize()

```python
text = "python PROGRAMMING"
print(text.capitalize())
```

#### title()

```python
text = "python programming language"
print(text.title())
```

#### swapcase()

```python
text = "PyThOn"
print(text.swapcase())
```

### Checking functions

#### islower()

```python
text = "python"
print(text.islower())
```

#### isupper()

```python
text = "PYTHOn"
print(text.isupper())
```

#### isalnum()

```python
text = "Python123"
print(text.isalnum())
```

#### isdigit()

```python
text = "12345"
print(text.isdigit())
```

### Strip()

```python
#strip(removes spaces from both sides)
text = "   Python   "
print(text.strip())

#lstrip()
text = "   Python   "
print(text.lstrip())

#rstrip()
text = "   Python   "
print(text.rstrip())
```

### Split()

```python
text = "Python is easy"
result = text.split()
print(result)

#Using Seperator
text = "apple,banana,mango"
result = text.split("->")
print(result)
```

### Join()

```python
words = ["Python", "is", "easy"]
result = " ".join(words)
print(result)

#Using seperator
words = ["Python", "Java", "C"]
print("-".join(words))
```

### Replace()

```python
text = "I like Java"
result = text.replace("Java", "Python")
print(result)

#Replacing multiple occurences
text = "apple apple apple"
print(text.replace("apple", "mango"))
```
