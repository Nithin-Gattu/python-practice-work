## Sets

```python
numbers = {10, 20, 30, 40}
print(numbers)
print(type(numbers))

# Duplicate values
numbers = {10, 20, 20, 30, 30, 40}
print(numbers)

# Empty Set
numbers = set()
print(numbers)
print(type(numbers))
```

### Set Operations 

#### Union (|)

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A | B) #using OR operator
print(A.union(B)) #using Union method
```

#### Intersection (&)

```python
print(A & B) #using AND operator
print(A.intersection(B)) #using intersection method
```

#### Difference (-)

```python
print(A - B) #using operator
print(B - A) #using Difference method
```

#### Symmetric Difference

```python
print(A ^ B) #using operator
print(A.symmetric_difference(B)) #using Symmetric Difference method
```

#### issubset()

```python
B = {2, 3, 4}
A = {1, 2, 3, 4, 5}
print(B.issubset(A))
```

#### issuperset()

```python
print(A.issuperset(B))
```

### Set Methods

#### add()

```python
numbers = {10, 20, 30}
numbers.add(40)
print(numbers)
```

#### update()

```python
numbers = {10, 20, 30}
numbers.update([40, 50, 60])
print(numbers)
```

#### remove()

```python
numbers = {10, 20, 30, 40}
numbers.remove(30)
numbers.remove(50) #error will display
print(numbers)
```

#### discard()

```python
numbers = {10, 20, 30, 40}
numbers.discard(30)
numbers.discard(50) #no error will display
print(numbers)
```

#### pop()

```python
numbers = {10, 20, 30, 40, 50}
x = numbers.pop()
print(x)
print(numbers)
```

#### clear()

```python
numbers = {10, 20, 30}
numbers.clear()
print(numbers)
```

## Dictionaries

```python
student = {
    "name": "Nithin",
    "age": 21,
    "course": "B.Tech"
}
print(student)
```

```python
# accessing values
student = {
    "name": "Nithin",
    "age": 21
}
print(student["name"])
print(student["age"])

print(student.get("name")) #using get()
```

```python
#adding elements
student = {
    "name": "Nithin",
    "age": 21
}
student["course"] = "B.Tech"
print(student)

#updating elements
student = {
    "name": "Nithin",
    "age": 21
}
student["age"] = 22
print(student)

#removing elements
student = {
    "name": "Nithin",
    "age": 21,
    "course": "B.Tech"
}
student.pop("age")
print(student)
```

### Dictionary Methods

#### keys()

```python
student = {
    "name": "Nithin",
    "age": 21,
    "course": "B.Tech"
}
print(student.keys())
```

#### values()

```python
print(student.values())
```

#### items()

```python
print(student.items())
```

#### get()

```python
print(student.get("name"))
print(student.get("salary")) #if key not present
```

#### update()

```python
student = {
    "name": "Nithin",
    "age": 21
}
student.update({"course": "B.Tech", "age": 22})
print(student)
```

#### clear()

```python
student = {
    "name": "Nithin",
    "age": 21
}
student.clear()
print(student)
```

### Dictionary Built-in Functions

#### len()

```python
student = {
    "name": "Nithin",
    "age": 21,
    "course": "B.Tech"
}
print(len(student))
```

#### min() & max()

```python
numbers = {
    10: "A",
    20: "B",
    5: "C"
}
print(min(numbers)) #min

print(max(numbers)) #max
```
